<h3><i>Dataset Overview</i></h3></div>

<div style="color: #14368B;">
    <p ><i><em></em>A real-world courier shipment dataset extracted from DTDC operations across India. This dataset captures the end-to-end flow of parcel logistics, covering sender/receiver details, weight, delivery mode, charges, and status.</em></i></p>
<p><em><i>Each row represents a courier consignment booked via DTDC, including details like:</em></i></p>
    <ul><li><em><i>Pickup and delivery location (city, pincode, state)</em></i></li>
    <li><em><i>Shipment weight (actual, volumetric, chargeable)</em></i></li>
    <li><em><i>Delivery mode and tariff breakdown</em></i></li>
    <li><em><i>Signatures, timestamps, and GST details (where applicable)</em></i></li>
    <li><em><i>Nature of parcel: documents or non-docs</em></i></li>
    </ul></em></i></p></div>

<div style="color: #14368B;">
    <p><em>The data set contains 42 columns</em></p>
<ul>
  <li><strong>Origin:</strong> <em>The city from which the parcel was originally booked or dispatched.</em></li>
  <li><strong>Destination:</strong> <em>The city where the parcel is intended to be delivered.</em></li>
  <li><strong>Pouch No:</strong> <em>A unique internal identifier (often UUID) used for grouping shipments in logistics.</em></li>
  <li><strong>Date:</strong> <em>The date when the shipment was booked, represented in Excel serial number format.</em></li>
  <li><strong>Sender Name:</strong> <em>Full name of the individual or company sending the parcel.</em></li>
  <li><strong>Sender Phone:</strong> <em>Mobile or landline contact number of the sender.</em></li>
  <li><strong>Sender Address:</strong> <em>Complete address of the sender for pickup and records.</em></li>
  <li><strong>Sender City:</strong> <em>The city of the sender may duplicate the "Origin" field.</em></li>
  <li><strong>Sender State:</strong> <em>The Indian state in which the sender is located.</em></li>
  <li><strong>Sender Pincode:</strong> <em>The 6-digit postal code (ZIP) of the sender's location.</em></li>
  <li><strong>Sender GSTIN:</strong> <em>GST Identification Number of the sender (if applicable for business shipments).</em></li>
  <li><strong>Total Pieces:</strong> <em>The number of individual physical packages included in the consignment.</em></li>
  <li><strong>Actual Wt:</strong> <em>The actual physical weight (in kilograms) of the parcel(s).</em></li>
  <li><strong>Volumetric Wt:</strong> <em>The weight calculated based on volume dimensions, used for billing purposes.</em></li>
  <li><strong>Chargeable Wt:</strong> <em>The weight used for billing , typically the greater of actual or volumetric weight.</em></li>
  <li><strong>Paperwork:</strong> <em>Indicates whether paperwork was submitted along with the parcel (Yes/No).</em></li>
  <li><strong>Sender Signature:</strong> <em>Denotes whether the sender signed the dispatch documentation (Yes/No).</em></li>
  <li><strong>Sender Date:</strong> <em>The date the sender completed handoff or signed documents, in Excel serial format.</em></li>
  <li><strong>Recipient Name:</strong> <em>The full name of the person or organization meant to receive the parcel.</em></li>
  <li><strong>Recipient Phone:</strong> <em>Contact number of the intended recipient.</em></li>
  <li><strong>Recipient Address:</strong> <em>The complete delivery address where the parcel should be sent.</em></li>
  <li><strong>Recipient City:</strong> <em>The city of the recipient may duplicate the "Destination" field.</em></li>
  <li><strong>Receiver State:</strong> <em>The Indian state where the recipient resides.</em></li>
  <li><strong>Receiver Pincode:</strong> <em>The 6-digit postal or ZIP code of the delivery address.</em></li>
  <li><strong>Description:</strong> <em>A short textual description of the parcel contents (e.g., documents, items).</em></li>
  <li><strong>Value Added Services:</strong> <em>Optional services added to the shipment, such as fragile handling, express delivery, insurance, etc.</em></li>
  <li><strong>Consignment No:</strong> <em>The official DTDC tracking number assigned to the shipment.</em></li>
  <li><strong>Expiry Date:</strong> <em>The last valid date for delivery, after which the shipment may expire or be returned. Stored as an Excel serial number.</em></li>
  <li><strong>Booking Code:</strong> <em>The unique code identifying the DTDC branch or booking counter.</em></li>
  <li><strong>Recipient GSTIN:</strong> <em>GST Identification Number of the recipient (if applicable).</em></li>
  <li><strong>Receiver Name:</strong> <em>An alternate or duplicate field for the recipient name.</em></li>
  <li><strong>Relationship:</strong> <em>Describes the relationship of the receiver to the original recipient (e.g., Self, Staff, Family).</em></li>
  <li><strong>Company Stamp:</strong> <em>Indicates whether a company stamp was used during the receipt (Yes/No).</em></li>
  <li><strong>Receiver Signature:</strong> <em>Indicates whether the recipient signed the delivery confirmation (Yes/No).</em></li>
  <li><strong>Receive Date:</strong> <em>The date the parcel was received, in Excel serial format.</em></li>
  <li><strong>Tariff:</strong> <em>The base delivery charge in Indian Rupees.</em></li>
  <li><strong>VAS Charges:</strong> <em>Additional charges for any value-added services applied to the shipment.</em></li>
  <li><strong>Total Amount:</strong> <em>The total amount billed for the consignment, including tariff and any add-ons.</em></li>
  <li><strong>Mode:</strong> <em>The mode of shipment chosen :Surface, Air, or Express.</em></li>
  <li><strong>Risk Surcharge:</strong> <em>Denotes who bears risk liability, either the Owner (sender) or Carrier (DTDC).</em></li>
  <li><strong>Mode of Payment:</strong> <em>The payment method used at booking, such as Cash, Card, UPI, Wallet, etc.</em></li>
  <li><strong>Nature of Consignment:</strong> <em>Indicates the type of parcel "Dox" for documents or "Non-Dox" for goods/merchandise.</em></li>
</ul>
</div>

<div style="color: #14368B;">
<p><i>Outlier checking</i></p>
</div>

```python
df[['Actual Wt', 'Volumetric Wt', 'Chargeable Wt', 'Tariff', 'VAS Charges', 'Total Amount']].describe()
```

<div style="color: #14368B;">
    <p><i>The mean and median values of all columns are almost same and comparing all values it seems to have an evenly distributed data.</i></p>
    <p><i>Only the tariff data seems to have a high max value almost double but that could be geniune large parcel.</i></p>
    <p><i>We can check that with IQR method </i></p>
</div>

```python
Q1 = df["Tariff"].quantile(0.25)
Q3 = df["Tariff"].quantile(0.75)
IQR = Q3-Q1
lowerBand = Q1 - 1.5 * IQR
upperBand = Q3 + 1.5 * IQR
print(f"Upper limit is {upperBand}")
print(f"Lower limit is {lowerBand}")
outliers = df[(df['Tariff'] < lowerBand) | (df['Tariff'] > upperBand)]
print(outliers.shape)
```

<div style="color: #14368B;">
    <p><i>Note : since the lower band is negative, no rows are under the lower limit</i></p>
    <p><i>The upper limit of 1112 rows could be actual parcels with high volumetric weight.</i></p>
    <p><i>To confirm that check the .describe of all coulmns for tariff greater than the upper limit alone</i></p>
</div>

```python
df[df["Tariff"]>1046.36].sort_values(by ="Tariff",ascending  = False)[['Chargeable Wt', 'Mode', 'Origin', 'Destination', 'Total Amount']]
```
