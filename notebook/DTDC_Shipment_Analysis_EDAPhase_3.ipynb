### 1. Removing PII columns

* **Columns removed:** `"Sender Phone"`, `"Sender's Name"`, `"Sender Address"`, `"Recipient Name"`, `"Recipient Phone"`, `"Recipient Address"`, `"Sender Pincode"`, `"Receiver Pincode"`

These columns were removed because they contain **personally identifiable information (PII)** and do not contribute directly to the shipment-level operational analysis.


```python
df = df.drop(columns = ["Sender Phone","Sender's Name","Sender Address","Recipient Name","Recipient Phone","Recipient Address","Sender Pincode","Receiver Pincode"])
```

```python
df.shape # checking the coulmn count after deletion of columns.
```

### 2. Converting String Date Columns to Datetime

* **Columns targeted:** `"Date"`, `"Receive Date"`, `"Sender Date"`, `"Expiry Date"`

These columns were converted from **string objects to datetime formats** to enable time-based operations, and trend analysis.


```python
df["Date"] = pd.to_datetime(df["Date"])
df["Receive Date"] = pd.to_datetime(df["Receive Date"])
df["Sender Date"] = pd.to_datetime(df["Sender Date"])
df["Expiry Date"] = pd.to_datetime(df["Expiry Date"])
```

### 3. Missing Value Handling

* **Columns analyzed:** `"Sender GSTIN"`, `"Sender Signature"`, `"Value Added Services"`, `"Recipient GSTIN"`, `"Company Stamp"`, `"Receiver Signature"`

#### Key Insights & Strategy:
* **Sender GSTIN & Sender Signature:** A cross-tabulation comparison reveals a strong correlation between the two. The signature is always present when a GSTIN is provided. Since unique values are limited to `"Yes"` or null, this functions as a binary **Yes/No** flag. Because GSTIN primarily applies to business entities rather than individuals, missing values are retained as `'Not applied'` instead of being treated as errors or missing data.
* **Recipient GSTIN & Receiver Signature:** These follow the exact same structural logic as the sender-side columns.
* **Company Stamp:** This field is sometimes present alongside a receiver signature but is not mandatory. Missing values will be filled with `'No'`.


```python
df["Sender Signature"].isna().sum()
```

```python
# checking values in the columns to confirm the existing values support the column as Yes/No 
print(df['Sender Signature'].unique())
print(df['Receiver Signature'].unique())
print(df['Company Stamp'].unique()) 
```

```python
pd.crosstab(df['Company Stamp'].isna(), df['Receiver Signature'].isna())
```

```python
df.loc[df["Value Added Services"].isna(),"VAS Charges"].sum()
```

* **Logic:** Cross-referenced `"Value Added Services"` with `"VAS charges"`. 
* **Finding:** For all rows where the service name is null, the corresponding charges sum up to exactly **zero**.

### 5.3. Outlier checking

```python
df[['Actual Wt', 'Volumetric Wt', 'Chargeable Wt', 'Tariff', 'VAS Charges', 'Total Amount']].describe()
```

* The mean and median values of all columns are almost the same. Comparing all values, the data seems to be evenly distributed.
* Only the tariff data seems to have a high maximum value (almost double), but that could be a genuine large parcel.
* We can check that with the IQR method.


```python
Q1 = df["Tariff"].quantile(0.25)
Q3 = df["Tariff"].quantile(0.75)
IQR = Q3-Q1
lowerBand = Q1 - 1.5 * IQR
upperBand = Q3 + 1.5 * IQR
print(f"Upper limit is {upperBand}")
print(f"Lower limit is {lowerBand}")
outliers = df[(df['Tariff'] < lowerBand) | (df['Tariff'] > upperBand)]
print(outliers.shape)
```

> **Note:** Since the lower band is negative, no rows fall below the lower limit.

* The IQR method identifies **1,112 high-tariff observations** as statistical outliers. 
* Inspection shows that these observations are associated with relatively high chargeable weights. Therefore, they appear to represent legitimate high-value shipments rather than obvious data-entry errors and were retained.

### 6.1 Risk Allocation Analysis

#### 1. Who typically bears shipment risk?

```python
riskCarrier_Count = eda_df["Risk Surcharge"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(riskCarrier_Count,labels = riskCarrier_Count.index , colors = ['#D8232A','#14368B'],autopct='%1.1f%%')
plt.title("Risk distribution")
plt.show()
```

**Insight:** 98.9% of shipments have the Carrier bearing the risk surcharge; only 1.1% fall to the Owner.
This heavy skew suggests that carrier risk is either the default (or mandatory) option, or one that customers overwhelmingly choose, which shifts liability for loss or damage to DTDC.

#### 2. Does the type of consignment affect who bears the risk?

*Note: Dox = documents; Non-Dox = parcels.*

```python
pd.crosstab(eda_df["Nature of Consignment"], eda_df["Risk Surcharge"])
```

```python
eda_df.pivot_table(
    index='Nature of Consignment',
    columns='Risk Surcharge',
    aggfunc='size'
)
```

**Insight:** Owner-borne risk occurs only for Dox shipments (547 cases); Non-Dox shipments are always Carrier-borne (0% Owner).

Risk Allocation by Consignment Type: Cross-tabulation reveals that risk allocation is strictly governed by the product category. Non-document consignments (parcels) are defaulted to Carrier Risk (100%), meaning users have no alternative but to transfer liability to DTDC. Conversely, Document-type consignments offer risk flexibility, representing the only segment where owners can choose who bears the financial liability for loss or damage.

#### 3. Does adding a Value Added Service change this pattern further?

```python
eda_df.pivot_table(
    index=['Nature of Consignment','Value Added Services'],
    columns='Risk Surcharge',
    aggfunc='size'
)
```

**Insight:** Owner bears risk only in a very specific combination — Document AND no VAS selected. Any VAS selection, or any Non-Document shipment, shifts risk entirely to the Carrier.

Risk distribution by Consignment type and Value Added Services: The above pivot -table reveals that owner bears the risk surcharge only when Nature of Consignment =  Document AND Value Added Services = None (not selected). 
In every other combination (any VAS selected, or any Non-Dox shipment), the Carrier always bears the risk. 
This could also imply that the Value Added Services (COD, Express, Insurance) include the risk allocation to carrier as mandatory.

#### 3. How does weight affect tariff?

```python
eda_df['Chargeable Wt'].corr(eda_df['Tariff'])
```

```python
numeric_cols = ['Actual Wt', 'Volumetric Wt', 'Chargeable Wt', 'Tariff', 'VAS Charges', 'Total Amount']
plt.figure(figsize=(8,6))
sns.heatmap(eda_df[numeric_cols].corr(), annot=True, fmt='.2f', cmap='Reds')
plt.title('Correlation Matrix — Numeric Columns')
plt.show()
```

```python
(eda_df['Chargeable Wt'] == eda_df['Volumetric Wt']).mean() * 100
```

**Insight:** Chargeable Weight equals Volumetric Weight in 100% of shipments (volumetric weight is higher than actual weight in every row), so pricing here follows package size, not physical weight.

**Insight:** The correlation matrix shows a clear chain of cost drivers. The three weight measures move almost in lockstep (Actual vs Volumetric Wt: 0.97; Volumetric vs Chargeable Wt: ~1.00). Weight feeds into Tariff (0.80-0.82), and Tariff in turn almost entirely determines Total Amount (0.98). VAS Charges, by contrast, are essentially unrelated to weight or Tariff (0.01) and only weakly related to Total Amount (0.20), so they work as an independent add-on rather than part of the weight-based pricing. In short: Weight → Tariff → Total Amount is the core pricing chain, and VAS is a small, separate layer on top.

#### 3.a. How do Mode and weight together determine tariff?

```python
plt.figure(figsize=(8,6))
colors = {'Surface': '#14368B', 'Air Cargo': '#F2A104', 'Express': '#D8232A'}
for mode, color in colors.items():
    subset = eda_df[eda_df['Mode'] == mode]
    plt.scatter(subset['Chargeable Wt'], subset['Tariff'], alpha=0.3, color=color, label=mode)

plt.title('Weight vs Tariff by Mode')
plt.xlabel('Chargeable Weight')
plt.ylabel('Tariff')
plt.legend()
plt.show()
```

**Insight:** While Chargeable Weight and Tariff show a strong overall correlation (0.82), coloring the scatter plot by Mode reveals the true structure: three distinct linear pricing bands, one per Mode. Express commands the steepest rate (highest Tariff per kg), followed by Air Cargo, with Surface having the lowest rate per kg. This means Mode doesn't just add a flat surcharge — it determines the entire pricing slope, confirming Mode as the primary driver of cost alongside Weight.

#### 4. How strongly does tariff drive total amount?

```python
plt.figure(figsize=(7,5))
plt.scatter(eda_df['Tariff'], eda_df['Total Amount'], alpha=0.3, color='#D8232A')
plt.title('Tariff vs Total Amount')
plt.xlabel('Tariff')
plt.ylabel('Total Amount')
plt.show()
```

```python
# Is Total Amount exactly Tariff + VAS Charges? (max absolute difference)
(eda_df['Total Amount'] - eda_df['Tariff'] - eda_df['VAS Charges']).abs().max()
```

**Insight:** Total Amount equals Tariff plus VAS Charges exactly, so the 0.98 correlation is partly arithmetic. It falls short of 1.0 only because VAS Charges vary between 0 and 150. Tariff, itself set by weight and Mode, drives almost the entire bill.

```python
vas_mode_pct = pd.crosstab(eda_df['Mode'], eda_df['Value Added Services'], normalize='index') * 100
vas_mode_long = vas_mode_pct.reset_index().melt(id_vars='Mode', var_name='VAS', value_name='Percentage')

fig = px.bar(
    vas_mode_long,
    x='Mode',
    y='Percentage',
    color='VAS',
    barmode='group',
    color_discrete_sequence=['#D8232A', '#14368B', '#F2A104', '#2E8B57'],
    title='VAS Selection Rate by Shipping Mode'
)
fig.update_layout(yaxis_title='Percentage of Shipments (%)', xaxis_title='Mode')
fig.show()
```

```python
eda_df['VAS Share of Total'] = (eda_df['VAS Charges'] / eda_df['Total Amount']) * 100
eda_df['VAS Share of Total'].describe()
```

**Step 2 – Only shipments that selected a VAS:** a more honest answer to "how much does VAS add when a customer actually chooses it?"

```python
eda_df[eda_df['Value Added Services'] != 'None']['VAS Share of Total'].describe()
```

```python
vas_only = eda_df[eda_df['Value Added Services'] != 'None']['VAS Share of Total']

plt.figure(figsize=(8,5))
plt.hist(vas_only, bins=30, color='#D8232A', edgecolor='black')
plt.axvline(vas_only.median(), color='#14368B', linestyle='--', linewidth=2,
            label=f'Median: {vas_only.median():.1f}%')
plt.title('VAS Share of Total Cost (Shipments with a VAS)')
plt.xlabel('VAS Share of Total Amount (%)')
plt.ylabel('Number of Shipments')
plt.legend()
plt.show()
```

**Step 3 – Checking the maximum (87%):** the highest VAS shares should come from very light shipments, where the base Tariff is tiny.

```python
eda_df.sort_values('VAS Share of Total', ascending=False)[['Chargeable Wt', 'Tariff', 'VAS Charges', 'Total Amount', 'VAS Share of Total']].head(5)
```

**Insight:** Among shipments that selected a Value Added Service, VAS typically adds about 17% to the total cost (median 16.8%; middle 50% range: 10.5%–25.3%). Across all shipments, including those with no VAS, the median falls to 13.4%. The extreme values (up to 87%) come from very light parcels, where a fixed VAS charge outweighs a small base Tariff. So VAS is a modest add-on for most shipments, and its share is largest on the lightest ones.

#### 1. How long do deliveries typically take?

```python
eda_df["Delivery Time"].unique()
```

```python
eda_df['Delivery Time'].value_counts().sort_index()
```

**Insight:** The delivery days varies from 1 day to 5 days. Highest number of shipments took 3 days of delivery days, but its not a high difference from other delivery days.

#### 2. Does shipping Mode affect delivery time?

```python
eda_df.groupby('Mode')['Delivery Time'].mean().sort_values() # for each mode what is the avg delivery time
```

```python
mode_delivery = eda_df.groupby('Mode')['Delivery Time'].mean().sort_values()

plt.figure(figsize=(7,5))
plt.bar(mode_delivery.index, mode_delivery.values, color=['#D8232A', '#14368B', '#F2A104'], edgecolor='black')
plt.title('Delivery Time Vs Mode')
plt.xlabel('Mode')
plt.ylabel('Delivery Time')
plt.ylim(0,5) 
plt.show()
```

**Insight:** Delivery time shows no meaningful relationship with shipping Mode. Express, Air Cargo and Surface all average ~3 days (2.99–3.01), even though Express costs about 77% more than Surface at the median (see Shipment Cost Analysis).

In this dataset Mode drives cost, not speed. That also fits Surface being the most-used mode: customers get the same delivery time for less. In a real operation this would be a red flag, because Express should deliver faster to justify its premium.

#### 3. Does the sender's state affect delivery time?

```python
state_delivery = eda_df.groupby('Sender State')['Delivery Time'].mean().sort_values(ascending=False)
state_delivery
```

```python
deliveryTime_state_mode = eda_df.pivot_table(values='Delivery Time', index='Sender State', columns='Mode', aggfunc='mean')

plt.figure(figsize=(8,10))
sns.heatmap(deliveryTime_state_mode, annot=True, fmt='.2f', cmap='Reds')
plt.title('Delivery Time by Mode StateWise')
plt.xlabel('Service Type')
plt.ylabel('Sender State')
plt.show()
```

Delivery time in this dataset appears independent of both Mode and State, likely a fixed/randomized range rather than an operationally-driven metric.
